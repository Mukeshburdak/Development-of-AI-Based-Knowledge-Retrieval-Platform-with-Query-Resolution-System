# Prompt: Create a simple student table in MySQL with student records and demonstrate basic INSERT, SELECT, WHERE, AND/OR, UPDATE, and DELETE operations.

```python

# -- Topic 1: CREATE TABLE --
create_table_sql = """
CREATE TABLE IF NOT EXISTS Students (
    StudentID INT PRIMARY KEY,
    FirstName VARCHAR(50),
    LastName VARCHAR(50),
    Age INT,
    Major VARCHAR(100),
    GPA DECIMAL(3, 2)
);
"""

# -- Topic 2: INSERT INTO --
insert_sql = """
INSERT INTO Students (StudentID, FirstName, LastName, Age, Major, GPA) VALUES
(1, 'Alice', 'Smith', 20, 'Computer Science', 3.8),
(2, 'Bob', 'Johnson', 21, 'Mathematics', 3.4),
(3, 'Charlie', 'Brown', 19, 'Physics', 2.9),
(4, 'Diana', 'Davis', 22, 'Computer Science', 3.95),
(5, 'Evan', 'Miller', 20, 'Chemistry', 3.1);
"""

# -- Topic 3: SELECT --
select_all_sql = """
SELECT * FROM Students;
"""

# -- Topic 4: WHERE --
where_sql = """
SELECT * FROM Students
WHERE GPA > 3.5;
"""

# -- Topic 5: AND, OR --
and_or_sql = """
SELECT * FROM Students
WHERE Major = 'Computer Science' AND GPA >= 3.8;
"""

# -- Topic 6: UPDATE --
update_sql = """
UPDATE Students
SET GPA = 3.5
WHERE StudentID = 2;
"""

# -- Topic 7: DELETE --
delete_sql = """
DELETE FROM Students
WHERE StudentID = 5;
"""

import sqlite3
import pandas as pd

conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

cursor.execute(create_table_sql)
cursor.execute(insert_sql)
print("--- Students Table ---")
print(pd.read_sql_query(select_all_sql, conn).to_string(index=False))
conn.close()
```

## Prompt: Demonstrate standard SQL Constraints including NOT NULL, UNIQUE, DEFAULT, CHECK, and FOREIGN KEY using SQLite.

```python
# -- Topic 24: SQL Constraints --
# This script demonstrates various SQL constraints including NOT NULL, UNIQUE, DEFAULT, CHECK, and FOREIGN KEY.

constraints_setup_sql = """
-- 1. Create a parent table with PRIMARY KEY and UNIQUE constraints
CREATE TABLE IF NOT EXISTS Departments (
    DeptID INTEGER PRIMARY KEY,
    DeptName TEXT NOT NULL UNIQUE
);

-- 2. Create a child table with NOT NULL, DEFAULT, CHECK, and FOREIGN KEY constraints
CREATE TABLE IF NOT EXISTS Employees (
    EmpID INTEGER PRIMARY KEY,
    FirstName TEXT NOT NULL,
    LastName TEXT NOT NULL,
    Email TEXT UNIQUE,
    Status TEXT DEFAULT 'Active',
    Salary REAL CHECK(Salary > 0),
    DeptID INTEGER,
    FOREIGN KEY (DeptID) REFERENCES Departments(DeptID)
);
"""

import sqlite3
import pandas as pd

# Establish an in-memory SQLite database connection
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# SQLite needs foreign key constraint enforcement explicitly enabled
cursor.execute("PRAGMA foreign_keys = ON;")

# Execute setup statements
cursor.executescript(constraints_setup_sql)

# Seed parent table
cursor.execute("INSERT INTO Departments (DeptID, DeptName) VALUES (10, 'Engineering'), (20, 'Human Resources');")

# Insert valid records
cursor.execute("""
INSERT INTO Employees (EmpID, FirstName, LastName, Email, Status, Salary, DeptID)
VALUES (1, 'Alice', 'Smith', 'alice@company.com', 'Active', 95000.0, 10);
""")

# Insert record relying on DEFAULT Status constraint
cursor.execute("""
INSERT INTO Employees (EmpID, FirstName, LastName, Email, Salary, DeptID)
VALUES (2, 'Bob', 'Jones', 'bob@company.com', 80000.0, 20);
""")

print("--- Departments Table ---")
df_depts = pd.read_sql_query("SELECT * FROM Departments;", conn)
display(df_depts)

print("\n--- Employees Table (Valid Records) ---")
df_emps = pd.read_sql_query("SELECT * FROM Employees;", conn)
display(df_emps)

# Demonstrate constraint violations gracefully using try-except blocks
print("\n--- Constraint Failure Demonstrations ---")

try:
    # Violating NOT NULL constraint
    cursor.execute("INSERT INTO Employees (EmpID, FirstName, LastName, Salary) VALUES (3, NULL, 'White', 50000.0);")
except sqlite3.IntegrityError as e:
    print(f"NOT NULL Violation Triggered: {e}")

try:
    # Violating UNIQUE constraint on Email
    cursor.execute("INSERT INTO Employees (EmpID, FirstName, LastName, Email, Salary) VALUES (4, 'John', 'Doe', 'alice@company.com', 60000.0);")
except sqlite3.IntegrityError as e:
    print(f"UNIQUE Violation Triggered: {e}")

try:
    # Violating CHECK constraint on Salary
    cursor.execute("INSERT INTO Employees (EmpID, FirstName, LastName, Salary) VALUES (5, 'Charlie', 'Brown', -100.0);")
except sqlite3.IntegrityError as e:
    print(f"CHECK Violation Triggered: {e}")

try:
    # Violating FOREIGN KEY constraint by passing non-existent DeptID
    cursor.execute("INSERT INTO Employees (EmpID, FirstName, LastName, Salary, DeptID) VALUES (6, 'Diana', 'Prince', 75000.0, 99);")
except sqlite3.IntegrityError as e:
    print(f"FOREIGN KEY Violation Triggered: {e}")

conn.close()
```

```python
# -- Topic 26: SQL User-Defined Functions --
# Standard SQLite does not support stored procedures (CREATE PROCEDURE).
# Instead, we register custom scalar functions using python's sqlite3 API and invoke them using standard SQL.

import sqlite3
import pandas as pd

# 1. Define a standard Python function to calculate grade status based on GPA
def get_gpa_status(gpa):
    if gpa >= 3.5:
        return "Honor Roll"
    elif gpa >= 3.0:
        return "Satisfactory"
    else:
        return "Needs Improvement"

# Establish connection
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# 2. Register our Python function as a SQL function named 'GPA_STATUS' taking 1 argument
conn.create_function("GPA_STATUS", 1, get_gpa_status)

# Create and populate Students table
cursor.execute("""
CREATE TABLE Students (
    StudentID INT PRIMARY KEY,
    FirstName VARCHAR(50),
    LastName VARCHAR(50),
    GPA DECIMAL(3, 2)
);
""")

cursor.execute("""
INSERT INTO Students (StudentID, FirstName, LastName, GPA) VALUES
(1, 'Alice', 'Smith', 3.8),
(2, 'Bob', 'Johnson', 3.4),
(3, 'Charlie', 'Brown', 2.9);
""")

# 3. Execute a SQL query utilizing our registered User-Defined Function
udf_sql = """
SELECT FirstName, LastName, GPA, GPA_STATUS(GPA) AS AcademicStatus
FROM Students;
"""

df_udf = pd.read_sql_query(udf_sql, conn)
display(df_udf)

conn.close()
```

## Prompt: Perform a database backup programmatically in SQLite using python's connection backup API.

```python
# -- Topic 27: Database Backup --
# Since SQLite does not have a standard 'BACKUP DATABASE' SQL command like other heavy engines,
# we use the standard Python SQLite3 backup API to clone our database cleanly into a physical file.

import sqlite3
import os

# 1. Establish an in-memory database and populate it with sample data
source_conn = sqlite3.connect(':memory:')
source_cursor = source_conn.cursor()

source_cursor.execute("""
CREATE TABLE IF NOT EXISTS SourceData (
    ID INTEGER PRIMARY KEY,
    Name TEXT NOT NULL
);
""")
source_cursor.execute("INSERT INTO SourceData (ID, Name) VALUES (1, 'Main Database Entry');")
source_conn.commit()

print("Source database initialized with sample records.")

# 2. Define our target backup file path
backup_file_path = "student_db_backup.db"

# Remove existing backup if it already exists
if os.path.exists(backup_file_path):
    os.remove(backup_file_path)

# 3. Establish a connection to the target backup database file
backup_conn = sqlite3.connect(backup_file_path)

# 4. Use python's sqlite3 backup API to copy the database dynamically and safely
with backup_conn:
    source_conn.backup(backup_conn)

print(f"Database backup completed successfully! Destination file: {backup_file_path}")

# 5. Verify the backup contains the correct data
verify_cursor = backup_conn.cursor()
verify_cursor.execute("SELECT * FROM SourceData;")
rows = verify_cursor.fetchall()
print(f"Verified backup content: {rows}")

# Clean up connections
backup_conn.close()
source_conn.close()
```

## Prompt: Compare the outputs of INNER JOIN, LEFT JOIN, and RIGHT JOIN side-by-side using a unified summary table.

```python
# -- Topic 30: Joins Comparison Summary --
import sqlite3
import pandas as pd

# Establish an in-memory SQLite database connection
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Re-create and populate the standard tables used in previous joins
cursor.execute("""
CREATE TABLE Departments (
    DeptID INT PRIMARY KEY,
    DeptName VARCHAR(100)
);
""")

cursor.execute("""
CREATE TABLE Students (
    StudentID INT PRIMARY KEY,
    FirstName VARCHAR(50),
    LastName VARCHAR(50),
    DeptID INT
);
""")

cursor.execute("""
INSERT INTO Departments (DeptID, DeptName) VALUES
(10, 'Computer Science'),
(20, 'Mathematics'),
(30, 'Physics'),
(40, 'Chemistry');
""")

cursor.execute("""
INSERT INTO Students (StudentID, FirstName, LastName, DeptID) VALUES
(1, 'Alice', 'Smith', 10),
(2, 'Bob', 'Johnson', 20),
(3, 'Charlie', 'Brown', 10),
(4, 'Diana', 'Davis', 30);
""")

# Define the three join queries
inner_join_query = """
SELECT Students.FirstName, Students.LastName, Departments.DeptName
FROM Students
INNER JOIN Departments ON Students.DeptID = Departments.DeptID;
"""

left_join_query = """
SELECT Students.FirstName, Students.LastName, Departments.DeptName
FROM Students
LEFT JOIN Departments ON Students.DeptID = Departments.DeptID;
"""

# Swapping the order of tables in LEFT JOIN to simulate RIGHT JOIN
right_join_query = """
SELECT Students.FirstName, Students.LastName, Departments.DeptName
FROM Departments
LEFT JOIN Students ON Departments.DeptID = Students.DeptID;
"""

# Fetch dataframes
df_inner = pd.read_sql_query(inner_join_query, conn)
df_left = pd.read_sql_query(left_join_query, conn)
df_right = pd.read_sql_query(right_join_query, conn)

# Display each join type outcome
print("=== 1. INNER JOIN RESULT ===\n(Only matching rows from BOTH tables)")
display(df_inner)

print("\n=== 2. LEFT JOIN RESULT ===\n(All rows from Students, with matched Departments)")
display(df_left)

print("\n=== 3. RIGHT JOIN RESULT ===\n(All rows from Departments, with matched Students)")
display(df_right)

conn.close()
```

## Prompt: Combine lists of students and employees into a single unified directory using the UNION operator, and filter the combined results to find people named 'Alice' using an outer query.

```python
# -- Topic 31: Filtering a UNION Query --
# To filter the combined results of a UNION, we wrap the entire UNION operation
# inside a subquery (or a Common Table Expression / CTE) and apply a WHERE clause on the outer query.

filtered_union_sql = """
SELECT *
FROM (
    SELECT FirstName, LastName, 'Student' AS Role FROM Students
    UNION
    SELECT FirstName, LastName, 'Employee' AS Role FROM Employees
) AS CombinedDirectory
WHERE FirstName = 'Alice';
"""

import sqlite3
import pandas as pd

# Establish an in-memory SQLite database connection
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create Tables
cursor.execute("""
CREATE TABLE Students (
    StudentID INT PRIMARY KEY,
    FirstName VARCHAR(50),
    LastName VARCHAR(50)
);
""")

cursor.execute("""
CREATE TABLE Employees (
    EmpID INT PRIMARY KEY,
    FirstName VARCHAR(50),
    LastName VARCHAR(50)
);
""")

# Populate Tables
cursor.execute("""
INSERT INTO Students (StudentID, FirstName, LastName) VALUES
(1, 'Alice', 'Smith'),
(2, 'Bob', 'Johnson'),
(3, 'Charlie', 'Brown');
""")

cursor.execute("""
INSERT INTO Employees (EmpID, FirstName, LastName) VALUES
(101, 'Alice', 'Cooper'),
(102, 'David', 'Miller'),
(103, 'Evan', 'Wright');
""")

# Run the filtered UNION query
df_filtered_union = pd.read_sql_query(filtered_union_sql, conn)
display(df_filtered_union)

conn.close()
```
