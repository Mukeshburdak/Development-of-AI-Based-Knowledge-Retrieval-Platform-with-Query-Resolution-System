## Prompt: Python - List Comprehension with simple examples of basic syntax, conditionals, and expressions

````python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]

# 1. Basic List Comprehension (Create a new list from an existing list)
# Syntax: [expression for item in iterable]
all_fruits = [x for x in fruits]
print("--- Basic Copy via Comprehension ---")
print(all_fruits)

# 2. Conditionals (Only accept elements that meet a condition)
# Syntax: [expression for item in iterable if condition == True]
# Example: Create a new list containing only fruits with the letter "a" in their name
newlist = [x for x in fruits if "a" in x]
print("\n--- Filtered List (fruits containing 'a') ---")
print(newlist)

# 3. Iterable (You can use any iterable object, like range())
# Example: Create a list of even numbers less than 10
even_numbers = [x for x in range(10) if x % 2 == 0]
print("\n--- List of Even Numbers via range() ---")
print(even_numbers)

# 4. Expression / Manipulation (Manipulate the value before it becomes a list item)
# Example: Set the values in the new list to upper case
upper_fruits = [x.upper() for x in fruits]
print("\n--- Upper Case Fruits ---")
print(upper_fruits)

# 5. Expression with Else condition
# Example: Replace "banana" with "orange"
replaced_list = [x if x != "banana" else "orange" for x in fruits]
print("\n--- Expression with Else ---")
print(replaced_list)
````

## Prompt: Python - Sort Lists with simple examples of alphabetical, numerical, descending, case-insensitive, and customized sorting using key functions

````python
# 1. Sort Alphabetically (Ascending by default)
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
thislist.sort()
print("--- Sort Alphabetically (Ascending) ---")
print(thislist)

# 2. Sort Numerically
numlist = [100, 50, 65, 82, 23]
numlist.sort()
print("\n--- Sort Numerically (Ascending) ---")
print(numlist)

# 3. Sort Descending (Using reverse = True)
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
thislist.sort(reverse = True)
print("\n--- Sort Alphabetically (Descending) ---")
print(thislist)

# 4. Customize Sort Function (Using a key function)
# Example: Sort the list based on how close the number is to 50
def myfunc(n):
  return abs(n - 50)

numlist = [100, 50, 65, 82, 23]
numlist.sort(key = myfunc)
print("\n--- Customize Sort (Closest to 50) ---")
print(numlist)

# 5. Case Insensitive Sort (By default, capital letters are sorted before lower case)
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort()
print("\n--- Case Sensitive Sort (Default) ---")
print(thislist)

# Perform a case-insensitive sort using str.lower as key
thislist.sort(key = str.lower)
print("\n--- Case Insensitive Sort ---")
print(thislist)
````

## Prompt: Python - Unpack Tuples with simple examples of basic unpacking and using the asterisk (*) operator

````python
# 1. Unpacking a Tuple
fruits = ("apple", "banana", "cherry")
(green, yellow, red) = fruits

print("--- Basic Unpacking ---")
print("green:", green)
print("yellow:", yellow)
print("red:", red)

# 2. Using Asterisk (*)
# If the number of variables is less than the number of values,
# you can add an '*' to the variable name and the values will be assigned to the variable as a list.
fruits_extended = ("apple", "banana", "cherry", "strawberry", "raspberry")
(green, yellow, *red) = fruits_extended

print("\n--- Unpacking with Asterisk (rest of items) ---")
print("green:", green)
print("yellow:", yellow)
print("red (list of remaining):", red)

# 3. Asterisk in the Middle
# If the asterisk is placed on another variable name than the last,
# Python will assign values to the variable until the number of values left matches the number of variables left.
(green, *tropic, red) = fruits_extended
print("\n--- Unpacking with Asterisk (middle items) ---")
print("green:", green)
print("tropic:", tropic)
print("red:", red)
````

##Prompt: Implement the linear regression model

```markdown
## Machine Learning: Simple Linear Regression

Linear Regression models the relationship between a dependent variable $y$ and an independent variable $x$ by fitting a linear equation to observed data.

$$
y = mx + c
$$

Where:
* **$y$** is the predicted output (Target)
* **$x$** is the input feature (Predictor)
* **$m$** is the slope (Coefficient/Weight)
* **$c$** is the y-intercept (Bias)
```

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# 1. Generate synthetic data with some noise
np.random.seed(42)
X = 2 * np.random.rand(100, 1)        # 100 features in a 2D array format
y = 4 + 3 * X + np.random.randn(100, 1)  # Target: y = 3x + 4 + Gaussian Noise

# 2. Instantiate and train (fit) the Linear Regression model
model = LinearRegression()
model.fit(X, y)

# 3. Extract the learned parameters
slope = model.coef_[0][0]
intercept = model.intercept_[0]

print("--- Model Parameter Results ---")
print(f"Estimated Slope (m):    {slope:.4f} (True Slope: 3)")
print(f"Estimated Intercept (c): {intercept:.4f} (True Intercept: 4)")
````

````python
# 4. Make predictions with our trained model
y_pred = model.predict(X)

# 5. Evaluate predictions using MSE and R2 Score
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print("--- Evaluation Metrics ---")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"R-squared (R2 Score):      {r2:.4f}")
````

````python
# 6. Visualize the data points and the regression line
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', alpha=0.6, label='Observed Data')
plt.plot(X, y_pred, color='red', linewidth=2, label=f'Fit Line: y = {slope:.2f}x + {intercept:.2f}')

plt.title("Simple Linear Regression Fit", fontsize=14, fontweight='bold')
plt.xlabel("Feature X")
plt.ylabel("Target y")
plt.legend(frameon=True, facecolor='white')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
````

```markdown
## Machine Learning: Logistic Regression (Classification)

Logistic Regression predicts the probability of a binary event occurring. It maps any real-valued number into a value between 0 and 1 using the **Sigmoid function**:

$$
S(z) = \frac{1}{1 + e^{-z}}
$$

Where $z = mx + c$ (similar to the linear regression equation). If the probability is $\geq 0.5$, we classify it as class `1`, otherwise as class `0`.
```

````python
# 5. Evaluate the classifier
accuracy = accuracy_score(y_test, y_pred)
print("--- Model Evaluation ---")
print(f"Accuracy Score: {accuracy:.4%}\n")
print("Classification Report:")
print(classification_report(y_test, y_pred))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
````

```markdown
A **Decision Tree** is a non-parametric supervised learning algorithm used for both classification and regression. It breaks down a dataset into smaller and smaller subsets while at the same time an associated decision tree is incrementally developed.

### Key Concepts:
* **Root Node:** The top-most decision node based on the most descriptive feature.
* **Decision Nodes:** Internal nodes where the data splits based on a feature condition (e.g., $Feature\ 1 \le 0.5$).
* **Leaf Nodes:** The end points of branches representing class predictions (no further splits).
* **Criteria (Gini / Entropy):** Metrics used to measure the quality of a split (minimizing impurity).
```

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# 1. Generate synthetic 2D dataset for classification
X_dt, y_class = make_classification(n_samples=100, n_features=2, n_redundant=0,
                                    n_informative=2, random_state=42, n_clusters_per_class=1)

# 2. Split into Train and Test sets
X_train, X_test, y_train, y_test = train_test_split(X_dt, y_class, test_size=0.3, random_state=42)

# 3. Instantiate and train a Decision Tree Classifier
# Setting max_depth=3 to prevent overfitting and make it easy to visualize
dt_clf = DecisionTreeClassifier(max_depth=3, random_state=42)
dt_clf.fit(X_train, y_train)

# 4. Make predictions and evaluate
y_pred_dt = dt_clf.predict(X_test)
print("--- Decision Tree Evaluation ---")
print(f"Accuracy Score: {accuracy_score(y_test, y_pred_dt):.2%}")
````

````python
# 5. Plot decision boundary along with the Tree structure
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Left subplot: Decision Boundary
x_min, x_max = X_dt[:, 0].min() - 0.5, X_dt[:, 0].max() + 0.5
y_min, y_max = X_dt[:, 1].min() - 0.5, X_dt[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02), np.arange(y_min, y_max, 0.02))

Z = dt_clf.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

ax1.contourf(xx, yy, Z, alpha=0.2, cmap='bwr')
ax1.scatter(X_dt[:, 0], X_dt[:, 1], c=y_class, cmap='bwr', edgecolors='k', alpha=0.8)
ax1.set_title("Decision Tree Boundary (depth=3)", fontsize=12, fontweight='bold')
ax1.set_xlabel("Feature 1")
ax1.set_ylabel("Feature 2")
ax1.grid(True, linestyle='--', alpha=0.5)

# Right subplot: Tree Diagram Structure
plot_tree(dt_clf, filled=True, feature_names=["Feature 1", "Feature 2"], class_names=["Class 0", "Class 1"], ax=ax2)
ax2.set_title("Learned Rule Tree structure", fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
````
